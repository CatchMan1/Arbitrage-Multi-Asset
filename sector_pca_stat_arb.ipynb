# 按行业导出 ETF 数据

从项目根目录运行。读取 `data/raw/etf_by_sector.xlsx` 的全部工作表，每个行业直接保存到 `data/etf_by_sector/`。保留原始列、行顺序、数值与代码字符串，不筛选或补齐数据，不重复保存另一套 CSV。

```python
from pathlib import Path
import pandas as pd

# 三个 Notebook 都从项目根目录运行，所有数据统一在 data 下
workbook_path = Path('data/raw/etf_by_sector.xlsx')
sector_dir = Path('data/etf_by_sector')
```

```python
# 一次读取全部行业；ETF 代码按字符串读取，保留前导零
sector_data = pd.read_excel(workbook_path, sheet_name=None, dtype={'code': str})
list(sector_data)
```

```python
# 每个工作表直接导出为同名 CSV，只保存这一份行业数据
sector_dir.mkdir(parents=True, exist_ok=True)

export_records = []
for sector, data in sector_data.items():
    csv_path = sector_dir / f'{sector}.csv'
    data.to_csv(csv_path, index=False, encoding='utf-8-sig')
    export_records.append({
        '行业': sector,
        '行数': len(data),
        'ETF数量': data['code'].nunique(),
        '导出文件': str(csv_path),
    })
```

```python
# 查看每个行业的导出行数、ETF 数量及保存位置
export_summary = pd.DataFrame(export_records)
export_summary
```
